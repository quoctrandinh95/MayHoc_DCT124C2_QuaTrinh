```python
import torch

a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print("a (torch.rand):\n", a)
print("b (torch.randn):\n", b)
print("c (torch.zeros):\n", c)
print("d (torch.ones):\n", d)
print("e (torch.eye):\n", e)

#Tensor 5x5 chứa toàn giá trị 10
f = torch.full((5, 5), 10)
print("\nTensor 5x5 toan gia tri 10 (torch.full):\n", f)
```

### Bài 6. Phép toán và nhân ma trận
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.  
Yêu cầu:   
• Phân biệt phép toán a * b với a @ b.T.   
• Cho biết shape của kết quả a @ b.T.

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\nMatrix multiplication (a @ b.T):")
res_matmul = a @ b.T
print(res_matmul)

print("\nElement-wise multiplication (a * b):")
res_elementwise = a * b
print(res_elementwise)
print("Shape của a * b:", res_elementwise.shape)

# Shape của kết quả a @ b.T
print("\nShape của a @ b.T:", res_matmul.shape)
```

Phân biệt phép toán a * b với a @ b.T:  
- a * b (Element-wise Multiplication - phép nhân theo từng phần tử): Nhân từng phần tử tại vị trí tương ứng giữa hai ma trận có cùng kích thước (2*3). Phần tử tại hàng i, cột j của kết quả bằng a_ij b_ij.  

- a @ b.T (Matrix Multiplication - phép nhân ma trận đại số tuyến tính): Thực hiện nhân ma trận $a$ kích thước (2*3) với ma trận chuyển vị b^T kích thước (3*2) theo quy tắc hàng nhân cột. 

### Bài 7. Các hàm thống kê trên Tensor
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.  
Yêu cầu:Thay dim=0 bằng dim=1 và nhận xét sự khác nhau.

```python
import torch

a = torch.randn(3, 4)

print("Original Tensor a:")
print(a)

print("\n--- THONG KE THEO COT (dim=0) ---")
print("Mean (dim=0):", torch.mean(a, dim=0))
print("Standard deviation (dim=0):", torch.std(a, dim=0))
print("Cumulative sum (dim=0):\n", torch.cumsum(a, dim=0))

# Yêu cầu: Thay dim=0 bằng dim=1
print("\n--- THONG KE THEO HANG (dim=1) ---")
print("Mean (dim=1):", torch.mean(a, dim=1))
print("Standard deviation (dim=1):", torch.std(a, dim=1))
print("Cumulative sum (dim=1):\n", torch.cumsum(a, dim=1))
```

Khi dim=0 (Tính toán dọc theo các hàng / gom theo từng cột):   
- Các phép toán thống kê thu gọn chiều dọc (chiều 0), giữ nguyên số cột.  
- torch.mean(a, dim=0) và torch.std(a, dim=0) trả về vector có kích thước torch.Size([4]), tương ứng với 4 giá trị trung bình/độ lệch chuẩn của 4 cột.  
- torch.cumsum(a, dim=0) tính tổng tích lũy dồn từ hàng trên xuống hàng dưới, kết quả giữ nguyên kích thước torch.Size([3, 4]). 
 
Khi dim=1 (Tính toán ngang qua các cột / gom theo từng hàng): 
- Các phép toán thống kê thu gọn chiều ngang (chiều 1), giữ nguyên số hàng.
- torch.mean(a, dim=1) và torch.std(a, dim=1) trả về vector có kích thước torch.Size([3]), tương ứng với 3 giá trị trung bình/độ lệch chuẩn của 3 hàng.
- torch.cumsum(a, dim=1) tính tổng tích lũy dồn từ cột bên trái sang cột bên phải của từng hàng, kết quả vẫn có kích thước torch.Size([3, 4]).

### Bài 8. Autograd cơ bản
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.
Yêu cầu:  
• Tính thủ công đạo hàm của y = x2 tại x = 3.6 và so sánh với x.grad.  
• Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

```python
import torch

# y = x * x
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()

print("--- HAM y = x * x ---")
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

# Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient
x_new = torch.tensor(3.6, requires_grad=True)
y_new = 3 * x_new**2 + 2 * x_new + 1
y_new.backward()

print("\n--- HAM y = 3*x^2 + 2*x + 1 ---")
print("x_new =", x_new)
print("y_new =", y_new)
print("x_new.grad =", x_new.grad)
```

Tính thủ công đạo hàm của y = x^2 tại x = 3.6 và so sánh với x.grad:  
- Hàm số: y = f(x) = x^2. 
- Đạo hàm lý thuyết theo giải tích: y' = f'(x) = dy/dx = 2x.  
- Thay giá trị tại x = 3.6:y'(3.6) = 2*3.6 = 7.2.  
- So sánh: Giá trị tính bằng tay trùng khớp với kết quả do PyTorch tự động tính (x.grad = tensor(7.2000)).

### Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.
Yêu cầu:  
• So sánh ba hệ số học được với [1, 2, 3].  
• Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().

```python
import numpy as np
import torch

# Sinh dữ liệu từ đa thức y = 1*x^2 + 2*x + 3
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

# Chuẩn bị ma trận đặc trưng [x^2, x, 1]
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

# Khởi tạo tham số và Tensor
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

# Vòng lặp huấn luyện Gradient Descent
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)
print(f"Final MSE Loss: {mse.item():.6f}")
```

So sánh ba hệ số học được với [1, 2, 3]:  
- Vector hệ số ban đầu khởi tạo ngẫu nhiên nên rất xa giá trị thực tế.  
- Sau 1000 bước lặp tối ưu bằng thuật toán NAdam, vector trọng số w hội tụ về các giá trị rất xấp xỉ bộ hệ số gốc [1, 2, 3] (thường đạt khoảng [1.000...,1.99...,2.9...]) với sai số toàn phương trung bình (MSE) tiệm cận về 0.  

Giải thích vai trò của 3 hàm trong vòng lặp huấn luyện:   
- optimizer.zero_grad(): Xóa sạch (reset) gradient tích lũy trong các biến w về 0 trước mỗi bước lặp mới. Trong PyTorch, mặc định các đạo hàm qua các lần gọi .backward() sẽ tự động cộng dồn lại; nếu không xóa, gradient sẽ bị sai lệch sau từng epoch.  
- mse.backward(): Kích hoạt cơ chế tự động tính đạo hàm (Autograd). Hàm này tính đạo hàm riêng của hàm mất mát mse theo từng tham số có requires_grad=True  sau đó lưu kết quả đạo hàm vào thuộc tính w.grad.
- optimizer.step(): Thực hiện cập nhật giá trị của trọng số w dựa trên đạo hàm vừa tính được trong w.grad theo quy tắc tối ưu hóa của thuật toán NAdam (tương tự việc dịch chuyển tham số ngược hướng dốc: w <- w - n.nabla_wmse).

#### Bài 10. Giải hệ phương trình bằng Autograd
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình. 
Yêu cầu:   
• Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.   
• Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
import random
import torch

# Khởi tạo ngẫu nhiên 4 ẩn số có hỗ trợ tính gradient
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

# Thiết lập tối ưu hóa NAdam
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

# Vòng lặp tối ưu tổng bình phương sai số
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("--- Nghiem tim duoc ---")
print(f"A = {A.item():.4f}")
print(f"B = {B.item():.4f}")
print(f"C = {C.item():.4f}")
print(f"D = {D.item():.4f}")
print(f"Tong sai so binh phuong (sqerr): {sqerr.item():.6e}")

# Kiểm tra nghiệm với 4 phương trình
print("\n--- Kiem tra 4 phuong trinh ---")
print(f"PT1: A + B = { (A + B).item():.4f} (Ky vong: 9)")
print(f"PT2: C - D = { (C - D).item():.4f} (Ky vong: 1)")
print(f"PT3: A + C = { (A + C).item():.4f} (Ky vong: 8)")
print(f"PT4: B - D = { (B - D).item():.4f} (Ky vong: 2)")
```

Kiểm tra nghiệm thu được với 4 phương trình:
- Thay giá trị tìm được (ví dụ một nghiệm điển hình A=4.72, B=4.28, C=3.28, D=2.28) vào hệ:  
   - A + B = 4.72 + 4.28 = 9.00 (thỏa mãn)  
   - C - D = 3.28 - 2.28 = 1.00 (thỏa mãn)  
   - A + C = 4.72 + 3.28 = 8.00 (thỏa mãn)  
   - B - D = 4.28 - 2.28 = 2.00 (thỏa mãn)  
- Tổng sai số sqerr đạt mức xấp xỉ $0$, chứng tỏ các giá trị học được đã giải quyết chính xác hệ bài toán.  

Nhận xét vì sao bài toán có thể có nhiều nghiệm:  
- Xét mặt đại số tuyến tính, hệ gồm 4 ẩn (A, B, C, D) nhưng 4 phương trình này phụ thuộc tuyến tính lẫn nhau:  
   - Lấy PT1 - PT2 + PT4 = (A + B) - (C - D) + (B - D) = A + 2B - C  
   - Hoặc biến đổi từ PT3 và PT4 : A - B + (C + D) tương đương; cụ thể: (PT1 + PT2) - (PT3 + PT4) = (A + B + C - D) - (A + C + B - D) = 0 = (9 + 1) - (8 + 2) = 0.
- Phương trình thứ tư là một hệ quả tuyến tính từ ba phương trình trước, do đó hạng ma trận hệ số chỉ bằng 3 (rank = 3 < 4 ẩn).
- Hệ phương trình có vô số nghiệm phụ thuộc vào một biến tự do (ví dụ đặt D = t, thì C = 1 + t, B = 2 + t, A = 7 - t). Tùy thuộc vào giá trị khởi tạo ngẫu nhiên ban đầu của A, B, C, D, thuật toán Gradient Descent sẽ hội tụ về nghiệm cục bộ gần điểm khởi tạo đó nhất.

### Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:

```python
import numpy as np
import torch

# YEU CAU 1: Tao Tensor X co 24 phan tu va reshape
print("=" * 60)
print("TAO VA RESHAPE TENSOR X")
print("=" * 60)

# Tao Tensor 1D chua 24 phan tu tu 1 den 24
X = torch.arange(1, 25)
print("Tensor X ban dau (1D, 24 phan tu):")
print(X)
print("Shape ban dau:", X.shape)

X_4x6 = X.reshape(4, 6)
X_2x12 = X.reshape(2, 12)
X_2x3x4 = X.reshape(2, 3, 4)

print("\n1. Reshape 4 x 6:\n", X_4x6, "\nShape:", X_4x6.shape)
print("\n2. Reshape 2 x 12:\n", X_2x12, "\nShape:", X_2x12.shape)
print("\n3. Reshape 2 x 3 x 4:\n", X_2x3x4, "\nShape:", X_2x3x4.shape)


# YEU CAU 2: Thao tac tren Tensor A, B kich thuoc 3 x 4
print("\n" + "=" * 60)
print("PHEP TOAN VA THONG KE TREN TENSOR A, B (3 x 4)")
print("=" * 60)

# Khoi tao Tensor A va B ngau nhien (hoac gia tri duong de phep chia ro rang)
torch.manual_seed(42)
A = torch.randint(1, 10, (3, 4), dtype=torch.float32)
B = torch.randint(1, 10, (3, 4), dtype=torch.float32)

print("Tensor A:\n", A)
print("Tensor B:\n", B)

print("\n1. A + B:\n", A + B)
print("\n2. A * B (Element-wise):\n", A * B)
print("\n3. A / B:\n", A / B)

# Tinh theo cot tuong ung voi dim=0
print("\n4. Trung binh theo cot (dim=0):")
print("   - Cua A:", torch.mean(A, dim=0))
print("   - Cua B:", torch.mean(B, dim=0))

print("\n5. Do lech chuan theo cot (dim=0):")
print("   - Cua A:", torch.std(A, dim=0))
print("   - Cua B:", torch.std(B, dim=0))


# YEU CAU 3: Tinh dao ham y = 2*x^2 + 5*x + 3 tai x = 2 bang autograd
print("\n" + "=" * 60)
print("YEU CAU 3: AUTOGRAD TINH DAO HAM TAI x = 2")
print("=" * 60)

x = torch.tensor(2.0, requires_grad=True)
y = 2 * (x**2) + 5 * x + 3
y.backward()

grad_autograd = x.grad.item()
print("Ham so: y = 2*x^2 + 5*x + 3")
print(f"Gia tri x = {x.item():.1f}, y = {y.item():.1f}")
print(f"Dao ham tinh boi PyTorch Autograd (dy/dx tai x = 2): {grad_autograd}")


# YEU CAU 4: Hoi quy da thuc y = 2*x^2 - 3*x + 5
print("\n" + "=" * 60)
print("YEU CAU 4: HOI QUY DA THUC y = 2*x^2 - 3*x + 5")
print("=" * 60)

# Sinh du lieu tu da thuc muc tieu: y = 2x^2 - 3x + 5
np.random.seed(42)
poly_target = np.poly1d([2, -3, 5])
N = 30
X_samples = np.random.randn(N, 1) * 3
Y_samples = poly_target(X_samples)

# Ma tran dac trung [x^2, x, 1]
XX = np.hstack([X_samples**2, X_samples, np.ones_like(X_samples)])
x_data = torch.tensor(XX, dtype=torch.float32)
y_data = torch.tensor(Y_samples, dtype=torch.float32)

# Tham so can toi uu w = [A, B, C]^T
w = torch.randn(3, 1, requires_grad=True)
optimizer = torch.optim.NAdam([w], lr=0.02)

print("He so khoi tao ban dau:")
print(w.detach().numpy().flatten())

for epoch in range(2500):
    optimizer.zero_grad()
    y_pred = x_data @ w
    loss = torch.mean((y_data - y_pred) ** 2)
    loss.backward()
    optimizer.step()

learned_coeffs = w.detach().numpy().flatten()
print("\nKet qua sau khi huan luyen:")
print(f"He so ky vong  : [A = 2.0, B = -3.0, C = 5.0]")
print(
    f"He so hoc duoc : [A = {learned_coeffs[0]:.4f}, B = {learned_coeffs[1]:.4f}, C = {learned_coeffs[2]:.4f}]"
)
print(f"Final MSE Loss : {loss.item():.6e}")
print("=" * 60)
```
